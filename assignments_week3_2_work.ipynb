## Data Analysis and Statistics Computation

```python
import pandas as pd

# Load cleaned teaching prices dataset
df = pd.read_csv('teaching_prices_clean.csv')
df.head()
```

```python
# Compute overall mean, median, std, min, and max for each item
summary_stats = df.groupby('item')['price'].agg(['mean', 'median', 'std', 'min', 'max'])
print(summary_stats)
```

```python
# Compute first-to-last percentage change for each item based on weekly mean
items = df['item'].unique()
results = []
for item in items:
    sub = df[df['item'] == item]
    date_means = sub.groupby('date')['price'].mean().sort_index()
    first_val = date_means.iloc[0]
    last_val = date_means.iloc[-1]
    pct_change = ((last_val - first_val) / first_val) * 100
    results.append({
        'item': item,
        'first_date_mean': first_val,
        'last_date_mean': last_val,
        'pct_change_%': round(pct_change, 2)
    })

pct_df = pd.DataFrame(results)
print(pct_df)
```
